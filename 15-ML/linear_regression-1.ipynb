```python
import numpy as np
X = np.array([2,3,4,6,10,3.5,10])
Y = np.array([35,45,60,75,90,75,30])
```

```python
X.shape
```

```python
X.ndim
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,Y)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()

model.fit(X.reshape(-1,1),Y)
```

```python
model.coef_[0]
```

```python
model.intercept_
```

```python
plt.figure(figsize=(10,5))
plt.scatter(X,Y)
plt.plot(X,model.predict(X.reshape(-1,1)),color='red')
plt.show()
```

```python
model.predict([[9.35]])
```
